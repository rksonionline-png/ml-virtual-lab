````python
from itertools import combinations

T = [{"Bread", "Milk"}, {"Bread", "Diaper", "Beer", "Eggs"}, {"Milk", "Diaper", "Beer", "Cola"},
     {"Bread", "Milk", "Diaper", "Beer"}, {"Bread", "Milk", "Diaper", "Cola"}]
N, min_count, min_conf = len(T), 3, 0.7

def count(s):
    return sum(1 for t in T if set(s) <= t)
````

### 1. Apriori: level-wise search with join and prune

````python
# 1. Apriori: level-wise search with join and prune
items = sorted({i for t in T for i in t})
L = [(i,) for i in items if count((i,)) >= min_count]
frequent = {s: count(s) for s in L}
k = 2
while L:
    cands = {tuple(sorted(set(a) | set(b))) for a in L for b in L if len(set(a) | set(b)) == k}
    cands = [c for c in cands if all(sub in frequent for sub in combinations(c, k - 1))]   # prune
    L = [c for c in sorted(cands) if count(c) >= min_count]
    frequent.update({c: count(c) for c in L})
    k += 1
for s, n in sorted(frequent.items(), key=lambda kv: (len(kv[0]), kv[0])):
    print(len(s), s, "count", n, "support", round(n / N, 2))
````

### 2. Rules from frequent itemsets

````python
# 2. Rules from frequent itemsets
print("rules (confidence >= 0.7):")
for s in [s for s in frequent if len(s) >= 2]:
    for r in range(1, len(s)):
        for A in combinations(s, r):
            B = tuple(i for i in s if i not in A)
            conf = frequent[s] / count(A)
            lift = conf / (count(B) / N)
            if conf >= min_conf:
                print(f"  {set(A)} -> {set(B)}  support={frequent[s] / N:.2f} confidence={conf:.2f} lift={lift:.2f}")
````
