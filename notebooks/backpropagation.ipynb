````python
import numpy as np

sig = lambda z: 1 / (1 + np.exp(-z))
````

### 1. Network 2 -> 2 -> 1, sigmoid units, loss L = 0.5 (o - t)^2

````python
# 1. Network 2 -> 2 -> 1, sigmoid units, loss L = 0.5 (o - t)^2
W = {"w11": .15, "w12": .20, "w21": .25, "w22": .30, "b1": .35, "b2": .35, "v1": .40, "v2": .45, "c": .60}
x1, x2, t = 0.05, 0.10, 0.01

def forward(W):
    h1 = sig(W["w11"] * x1 + W["w12"] * x2 + W["b1"])
    h2 = sig(W["w21"] * x1 + W["w22"] * x2 + W["b2"])
    o = sig(W["v1"] * h1 + W["v2"] * h2 + W["c"])
    return h1, h2, o, 0.5 * (o - t) ** 2

h1, h2, o, L = forward(W)
print(f"forward: h1 = {h1:.4f}, h2 = {h2:.4f}, o = {o:.4f}, loss = {L:.5f}")
````

### 2. Backward pass: error terms (delta) and gradients

````python
# 2. Backward pass: error terms (delta) and gradients
d_out = (o - t) * o * (1 - o)
d1 = d_out * W["v1"] * h1 * (1 - h1)
d2 = d_out * W["v2"] * h2 * (1 - h2)
g = {"v1": d_out * h1, "v2": d_out * h2, "c": d_out,
     "w11": d1 * x1, "w12": d1 * x2, "b1": d1, "w21": d2 * x1, "w22": d2 * x2, "b2": d2}
print(f"delta_out = {d_out:.5f}, delta_1 = {d1:.5f}, delta_2 = {d2:.5f}")
````

### 3. Gradient check against numerical derivatives

````python
# 3. Gradient check against numerical derivatives
for k in ("v1", "w11", "b2"):
    up, dn = dict(W), dict(W); up[k] += 1e-6; dn[k] -= 1e-6
    num = (forward(up)[3] - forward(dn)[3]) / 2e-6
    print(f"{k:4s} backprop = {g[k]: .8f}   numerical = {num: .8f}")
````

### 4. Gradient descent: the loss falls

````python
# 4. Gradient descent: the loss falls
eta = 0.5
for step in range(1, 10001):
    h1, h2, o, L = forward(W)
    d_out = (o - t) * o * (1 - o); d1 = d_out * W["v1"] * h1 * (1 - h1); d2 = d_out * W["v2"] * h2 * (1 - h2)
    g = {"v1": d_out * h1, "v2": d_out * h2, "c": d_out, "w11": d1 * x1, "w12": d1 * x2, "b1": d1, "w21": d2 * x1, "w22": d2 * x2, "b2": d2}
    for k in W:
        W[k] -= eta * g[k]
    if step in (1, 10, 100, 1000, 10000):
        print(f"step {step:5d}: loss = {L:.6f}, output = {o:.4f}")
````

### 5. The sigmoid derivative is at most 0.25 (vanishing gradients)

````python
# 5. The sigmoid derivative is at most 0.25 (vanishing gradients)
z = np.linspace(-6, 6, 1001)
print("max of sigma'(z) =", round(float((sig(z) * (1 - sig(z))).max()), 4))
````
