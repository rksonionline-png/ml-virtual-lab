````python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
````

### 1. Genre vectors: Action, Romance, Sci-Fi, Comedy, Drama

````python
# 1. Genre vectors: Action, Romance, Sci-Fi, Comedy, Drama
films = {"Star Raiders": [.9, 0, .9, 0, .1], "Love in Paris": [0, 1, 0, .2, .6], "Galaxy Quest 2": [.5, 0, .8, .7, 0],
         "Iron Fist": [1, 0, 0, 0, .2], "Laugh Factory": [0, .1, 0, 1, 0], "Tears of June": [0, .7, 0, 0, 1],
         "Mars Colony": [.3, 0, 1, 0, .5], "Wedding Mess": [0, .6, 0, .9, 0], "Night Chase": [.9, .1, 0, 0, .4],
         "Quiet Hearts": [0, .5, 0, 0, 1]}
names = list(films)
X = np.array(list(films.values()))
liked = ["Star Raiders", "Mars Colony"]
````

### 2. User profile = average of the liked items

````python
# 2. User profile = average of the liked items
profile = X[[names.index(n) for n in liked]].mean(axis=0)
print("profile (Action, Romance, SciFi, Comedy, Drama):", np.round(profile, 2))
````

### 3. Rank the unseen films by cosine similarity to the profile

````python
# 3. Rank the unseen films by cosine similarity to the profile
scores = cosine_similarity(profile[None, :], X)[0]
for s, n in sorted(zip(scores, names), reverse=True):
    if n not in liked:
        print(f"{n:15s} cosine = {s:.3f}")
````

### 4. Dot product favours long vectors, cosine does not

````python
# 4. Dot product favours long vectors, cosine does not
dots = X @ profile
print("top by dot product:", names[int(np.argmax([d if n not in liked else -1 for d, n in zip(dots, names)]))])
````

### 5. Text features with TF-IDF

````python
# 5. Text features with TF-IDF
docs = ["space ship battle against alien empire", "romantic story in paris spring", "robot crew explores mars colony",
        "funny wedding disaster comedy", "soldier fights in space war"]
tfidf = TfidfVectorizer().fit_transform(docs)
sim = cosine_similarity(tfidf[0], tfidf)[0]
print("similarity of the first description to all:", np.round(sim, 3))
````
