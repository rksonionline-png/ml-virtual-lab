````python
import numpy as np
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.datasets import load_iris

def entropy(y):
    _, counts = np.unique(y, return_counts=True)
    p = counts / counts.sum()
    return -np.sum(p * np.log2(p))

def gini(y):
    _, counts = np.unique(y, return_counts=True)
    p = counts / counts.sum()
    return 1 - np.sum(p ** 2)
````

### Parent node: 9 Yes, 5 No (play-tennis)

````python
# Parent node: 9 Yes, 5 No (play-tennis)
parent = np.array([1] * 9 + [0] * 5)
print("Entropy(parent) =", round(entropy(parent), 3))
print("Gini(parent)    =", round(gini(parent), 3))
````

### Split on Outlook: Sunny (2Y,3N), Overcast (4Y,0N), Rainy (3Y,2N)

````python
# Split on Outlook: Sunny (2Y,3N), Overcast (4Y,0N), Rainy (3Y,2N)
groups = [np.array([1]*2 + [0]*3), np.array([1]*4), np.array([1]*3 + [0]*2)]
weighted = sum(len(g) / len(parent) * entropy(g) for g in groups)
print("Information gain (Outlook) =", round(entropy(parent) - weighted, 3))

def best_split(X, y):
    best = (None, None, -1)
    for j in range(X.shape[1]):
        for t in np.unique(X[:, j]):
            left, right = y[X[:, j] <= t], y[X[:, j] > t]
            if len(left) == 0 or len(right) == 0:
                continue
            gain = gini(y) - (len(left) * gini(left) + len(right) * gini(right)) / len(y)
            if gain > best[2]:
                best = (j, t, gain)
    return best

iris = load_iris()
j, t, g = best_split(iris.data, iris.target)
print(f"Best root split: {iris.feature_names[j]} <= {t}  (Gini gain {g:.3f})")

clf = DecisionTreeClassifier(max_depth=2, random_state=0).fit(iris.data, iris.target)
print(export_text(clf, feature_names=iris.feature_names))
print("Training accuracy:", round(clf.score(iris.data, iris.target), 3))
````
