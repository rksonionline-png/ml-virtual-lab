````python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
````

### 1. Convolution, ReLU and max pooling in NumPy

````python
# 1. Convolution, ReLU and max pooling in NumPy
def conv2d(img, K, stride=1):
    k = K.shape[0]; n = (img.shape[0] - k) // stride + 1
    return np.array([[(K * img[i * stride:i * stride + k, j * stride:j * stride + k]).sum() for j in range(n)] for i in range(n)])

def max_pool(a):
    n = a.shape[0] // 2
    return a[:2 * n, :2 * n].reshape(n, 2, n, 2).max(axis=(1, 3))

img = np.zeros((14, 14))
img[3, 3:11] = img[10, 3:11] = img[3:11, 3] = img[3:11, 10] = 1           # a square outline
vertical = np.array([[-1, 0, 1]] * 3, dtype=float)
fmap = np.maximum(conv2d(img, vertical), 0)                                # convolution + ReLU
print("image 14x14 -> feature map", fmap.shape, " output size = (14 - 3)/1 + 1 =", (14 - 3) // 1 + 1)
print("columns with a response:", np.flatnonzero(fmap.sum(axis=0)))
print("after 2x2 max pooling:", max_pool(fmap).shape, " stride 2 conv:", conv2d(img, vertical, 2).shape)
````

### 2. Parameter counts

````python
# 2. Parameter counts
k, c_in, c_out = 3, 3, 16
print("conv layer parameters:", k * k * c_in * c_out + c_out)
print("dense layer 14x14 -> 12x12 would need", 14 * 14 * 12 * 12, "weights; the conv kernel needs 9")
````

### 3. A small neural network on 8x8 digit images

````python
# 3. A small neural network on 8x8 digit images
X, y = load_digits(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X / 16.0, y, test_size=0.3, random_state=0, stratify=y)
for sizes in [(), (32,), (64, 32)]:
    if sizes == ():
        from sklearn.linear_model import LogisticRegression
        m = LogisticRegression(max_iter=2000).fit(Xtr, ytr)
        name = "no hidden layer (logistic)"
    else:
        m = MLPClassifier(sizes, max_iter=500, random_state=0).fit(Xtr, ytr)
        name = f"hidden layers {sizes}"
    print(f"{name:28s} test accuracy = {m.score(Xte, yte):.3f}")
````
