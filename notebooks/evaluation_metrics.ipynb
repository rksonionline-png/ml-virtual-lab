````python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score, roc_auc_score, roc_curve
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
````

### 1. Metrics by hand from a confusion matrix

````python
# 1. Metrics by hand from a confusion matrix
TP, FP, FN, TN = 40, 10, 20, 130
print("accuracy", round((TP + TN) / (TP + FP + FN + TN), 3), " precision", round(TP / (TP + FP), 3),
      " recall", round(TP / (TP + FN), 3), " specificity", round(TN / (TN + FP), 3),
      " F1", round(2 * TP / (2 * TP + FP + FN), 3))
````

### 2. A real classifier on the breast-cancer data (class 1 = benign, class 0 = malignant)

````python
# 2. A real classifier on the breast-cancer data (class 1 = benign, class 0 = malignant)
X, y = load_breast_cancer(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)
clf = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(Xtr, ytr)
score = clf.predict_proba(Xte)[:, 1]
````

### 3. Move the threshold and watch precision and recall trade off

````python
# 3. Move the threshold and watch precision and recall trade off
for t in (0.1, 0.5, 0.9):
    pred = (score >= t).astype(int)
    tn, fp, fn, tp = confusion_matrix(yte, pred).ravel()
    print(f"t = {t}: TP={tp} FP={fp} FN={fn} TN={tn}  precision={precision_score(yte, pred):.3f}  recall={recall_score(yte, pred):.3f}  F1={f1_score(yte, pred):.3f}")
````

### 4. ROC and AUC

````python
# 4. ROC and AUC
fpr, tpr, thr = roc_curve(yte, score)
print("AUC =", round(roc_auc_score(yte, score), 4), " points on the ROC curve:", len(fpr))
````

### 5. Accuracy on imbalanced data can mislead

````python
# 5. Accuracy on imbalanced data can mislead
y_imb = np.array([1] * 5 + [0] * 95)
always_neg = np.zeros(100, dtype=int)
print("always-negative accuracy:", (always_neg == y_imb).mean(), " recall:", recall_score(y_imb, always_neg))
````
