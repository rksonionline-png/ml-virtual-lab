````python
import numpy as np
from scipy.cluster.hierarchy import linkage, fcluster
from sklearn.cluster import AgglomerativeClustering
````

### 1. Five points on a line, merged step by step

````python
# 1. Five points on a line, merged step by step
x = np.array([[1.0], [3.0], [4.0], [9.0], [11.0]])
for method in ("single", "complete", "average"):
    Z = linkage(x, method=method)
    print(method, "merge heights:", np.round(Z[:, 2], 2))
````

### 2. Reading the linkage matrix (complete linkage)

````python
# 2. Reading the linkage matrix (complete linkage)
Z = linkage(x, method="complete")
print("columns = cluster a, cluster b, height, size")
print(np.round(Z, 2))
````

### 3. Cut the tree into k clusters

````python
# 3. Cut the tree into k clusters
for k in (2, 3):
    print(k, "clusters:", fcluster(Z, t=k, criterion="maxclust"))
````

### 4. The same with scikit-learn

````python
# 4. The same with scikit-learn
rng = np.random.default_rng(0)
X = np.vstack([rng.normal([0, 0], 0.5, (10, 2)), rng.normal([4, 4], 0.5, (10, 2)), rng.normal([0, 5], 0.5, (10, 2))])
for link in ("single", "complete", "average", "ward"):
    lab = AgglomerativeClustering(n_clusters=3, linkage=link).fit_predict(X)
    print(f"{link:8s} cluster sizes:", np.bincount(lab))
````
