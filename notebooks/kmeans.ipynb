````python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
````

### 1. k-means from scratch (Lloyd's algorithm)

````python
# 1. k-means from scratch (Lloyd's algorithm)
def kmeans(X, k, iters=100, seed=0):
    rng = np.random.default_rng(seed)
    cent = X[rng.choice(len(X), k, replace=False)]
    for it in range(iters):
        d = ((X[:, None, :] - cent[None, :, :]) ** 2).sum(axis=2)   # distance to each centroid
        lab = d.argmin(axis=1)                                      # assignment step
        new = np.array([X[lab == j].mean(axis=0) if np.any(lab == j) else cent[j] for j in range(k)])
        if np.allclose(new, cent):                                  # update step
            break
        cent = new
    wcss = ((X - cent[lab]) ** 2).sum()
    return cent, lab, wcss, it + 1

X, _ = make_blobs(n_samples=150, centers=3, cluster_std=0.9, random_state=7)
cent, lab, wcss, n_it = kmeans(X, 3, seed=1)
print("centroids:\n", np.round(cent[np.argsort(cent[:, 0])], 2))
print("cluster sizes:", np.bincount(lab), "WCSS:", round(wcss, 2), "iterations:", n_it)
````

### 2. Compare with scikit-learn

````python
# 2. Compare with scikit-learn
km = KMeans(n_clusters=3, n_init=10, random_state=0).fit(X)
print("sklearn inertia:", round(km.inertia_, 2))
````

### 3. Elbow method

````python
# 3. Elbow method
for k in range(1, 7):
    w = KMeans(n_clusters=k, n_init=10, random_state=0).fit(X).inertia_
    print(f"k = {k}  WCSS = {w:8.2f}")
````
