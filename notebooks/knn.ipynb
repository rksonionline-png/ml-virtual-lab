````python
import numpy as np
from collections import Counter
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

X = np.array([[1, 1], [2, 1], [1.5, 2.5], [6, 5], [7, 6], [6.5, 7]], dtype=float)
y = np.array(["A", "A", "A", "B", "B", "B"])

def knn_predict(X, y, q, k=3):
    d = np.sqrt(((X - q) ** 2).sum(axis=1))
    idx = np.argsort(d)[:k]
    print("distances:", np.round(d, 2))
    print("nearest idx:", idx, "labels:", [str(v) for v in y[idx]])
    return Counter(y[idx]).most_common(1)[0][0]

print("Prediction for (3, 3):", knn_predict(X, y, np.array([3.0, 3.0]), k=3))
````

### Choosing k on Iris

````python
# Choosing k on Iris
iris = load_iris()
Xtr, Xte, ytr, yte = train_test_split(iris.data, iris.target, test_size=0.3,
                                      random_state=42, stratify=iris.target)
for k in (1, 3, 5, 9, 15, 31):
    acc = KNeighborsClassifier(n_neighbors=k).fit(Xtr, ytr).score(Xte, yte)
    print(f"k = {k:2d}  test accuracy = {acc:.3f}")
````
