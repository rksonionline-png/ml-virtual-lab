````python
import numpy as np
from sklearn.linear_model import LinearRegression
````

### Study hours (x) vs exam score (y)

````python
# Study hours (x) vs exam score (y)
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([52, 58, 65, 70, 78], dtype=float)
````

### From scratch: least squares formulas

````python
# --- From scratch: least squares formulas ---
x_bar, y_bar = x.mean(), y.mean()
m = np.sum((x - x_bar) * (y - y_bar)) / np.sum((x - x_bar) ** 2)
c = y_bar - m * x_bar
print(f"slope m = {m:.3f}, intercept c = {c:.3f}")

y_hat = m * x + c
sse = np.sum((y - y_hat) ** 2)
r2 = 1 - sse / np.sum((y - y_bar) ** 2)
print(f"SSE = {sse:.3f}, MSE = {sse/len(x):.3f}, R2 = {r2:.4f}")
print("Prediction for 6 hours:", round(m * 6 + c, 2))
````

### From scratch: gradient descent

````python
# --- From scratch: gradient descent ---
w, b, lr = 0.0, 0.0, 0.02
for epoch in range(2000):
    err = (w * x + b) - y
    w -= lr * (2 / len(x)) * np.sum(err * x)
    b -= lr * (2 / len(x)) * np.sum(err)
print(f"GD result  w = {w:.3f}, b = {b:.3f}")
````

### scikit-learn check

````python
# --- scikit-learn check ---
model = LinearRegression().fit(x.reshape(-1, 1), y)
print(f"sklearn    w = {model.coef_[0]:.3f}, b = {model.intercept_:.3f}")
````
