````python
import numpy as np
from sklearn.linear_model import LogisticRegression
````

### Hours studied -> pass (1) / fail (0)

````python
# Hours studied -> pass (1) / fail (0)
x = np.array([1, 2, 3, 4, 5, 6, 7, 8], dtype=float)
y = np.array([0, 0, 0, 1, 0, 1, 1, 1], dtype=float)   # overlap near x = 4-5

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

w, b, lr = 0.0, 0.0, 0.1
for epoch in range(1, 5001):
    p = sigmoid(w * x + b)
    w -= lr * np.mean((p - y) * x)      # dLoss/dw
    b -= lr * np.mean(p - y)            # dLoss/db
    if epoch in (1, 100, 1000, 5000):
        p = np.clip(sigmoid(w * x + b), 1e-12, 1 - 1e-12)
        loss = -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))
        print(f"epoch {epoch:5d}  loss = {loss:.4f}  w = {w:.3f}  b = {b:.3f}")

print("P(pass | 4.5 h) =", round(float(sigmoid(w * 4.5 + b)), 3))
print("Decision boundary x = -b/w =", round(-b / w, 2))

clf = LogisticRegression(C=1e6).fit(x.reshape(-1, 1), y)
print("sklearn P(pass | 4.5 h) =", round(float(clf.predict_proba([[4.5]])[0, 1]), 3))
print("sklearn w, b =", round(float(clf.coef_[0][0]), 3), round(float(clf.intercept_[0]), 3))
````
