````python
import numpy as np
from sklearn.model_selection import KFold, cross_val_score, GridSearchCV, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.svm import SVC
from sklearn.datasets import load_iris
````

### 1. Choose a polynomial degree by 5-fold cross-validation

````python
# 1. Choose a polynomial degree by 5-fold cross-validation
rng = np.random.default_rng(0)
x = rng.uniform(0, 10, 40)
y = 2.2 * np.sin(0.65 * x) + 0.25 * x - 1.2 + 0.8 * rng.normal(size=40)
X = x[:, None]
cv = KFold(5, shuffle=True, random_state=0)
for deg in range(1, 10):
    model = make_pipeline(PolynomialFeatures(deg), StandardScaler(), LinearRegression())
    s = -cross_val_score(model, X, y, cv=cv, scoring="neg_mean_squared_error")
    print(f"degree {deg}: CV MSE = {s.mean():8.3f} +- {s.std(ddof=1):6.3f}")
````

### 2. Ridge penalty chosen by cross-validation

````python
# 2. Ridge penalty chosen by cross-validation
for alpha in (0.001, 0.1, 1, 10):
    model = make_pipeline(PolynomialFeatures(9), StandardScaler(), Ridge(alpha=alpha))
    s = -cross_val_score(model, X, y, cv=cv, scoring="neg_mean_squared_error").mean()
    print(f"degree 9, ridge alpha = {alpha:6}: CV MSE = {s:8.3f}")
````

### 3. Grid search on Iris: count the fits

````python
# 3. Grid search on Iris: count the fits
iris = load_iris()
Xtr, Xte, ytr, yte = train_test_split(iris.data, iris.target, test_size=0.3, random_state=0, stratify=iris.target)
grid = {"svc__C": [0.1, 1, 10], "svc__gamma": [0.01, 0.1, 1]}
gs = GridSearchCV(make_pipeline(StandardScaler(), SVC()), grid, cv=5).fit(Xtr, ytr)
print("combinations:", 3 * 3, " folds: 5  ->", 3 * 3 * 5, "fits (+1 refit)")
print("best params:", gs.best_params_, " CV accuracy:", round(gs.best_score_, 3), " test accuracy:", round(gs.score(Xte, yte), 3))
````
