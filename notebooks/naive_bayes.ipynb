### Play-tennis style data: (Outlook, Windy) -> Play

````python
# Play-tennis style data: (Outlook, Windy) -> Play
data = [("Sunny","No","No"),("Sunny","Yes","No"),("Overcast","No","Yes"),
        ("Rainy","No","Yes"),("Rainy","No","Yes"),("Rainy","Yes","No"),
        ("Overcast","Yes","Yes"),("Sunny","No","Yes"),("Sunny","No","Yes"),
        ("Rainy","No","Yes"),("Sunny","Yes","Yes"),("Overcast","Yes","Yes"),
        ("Overcast","No","Yes"),("Rainy","Yes","No")]

labels = [r[-1] for r in data]
prior = {c: n / len(data) for c, n in Counter(labels).items()}
print("Priors:", {c: round(p, 3) for c, p in prior.items()})

def likelihood(col, value, cls, alpha=0):
    rows = [r for r in data if r[-1] == cls]
    k = len(set(r[col] for r in data))
    return (sum(1 for r in rows if r[col] == value) + alpha) / (len(rows) + alpha * k)

def posterior(query, alpha=0):
    score = {}
    for cls in prior:
        s = prior[cls]
        for col, v in enumerate(query):
            s *= likelihood(col, v, cls, alpha)
        score[cls] = s
    total = sum(score.values())
    return {c: s / total for c, s in score.items()}, score

query = ("Sunny", "Yes")
post, raw = posterior(query)
print("Unnormalised:", {c: round(s, 4) for c, s in raw.items()})
print("Posterior   :", {c: round(p, 3) for c, p in post.items()})
print("Prediction  :", max(post, key=post.get))
````

### Zero-frequency problem: Overcast & windy never occurs with 'No'

````python
# Zero-frequency problem: Overcast & windy never occurs with 'No'
post0, _ = posterior(("Overcast", "Yes"), alpha=0)
post1, _ = posterior(("Overcast", "Yes"), alpha=1)
print("No smoothing  P(No) =", round(post0["No"], 3))
print("Laplace (a=1) P(No) =", round(post1["No"], 3))
````
