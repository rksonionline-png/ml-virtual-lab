````python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
````

### 1. PCA from scratch on five points

````python
# 1. PCA from scratch on five points
X = np.array([[2, 3], [3, 5], [5, 4], [7, 8], [8, 9]], dtype=float)
Xc = X - X.mean(axis=0)
C = Xc.T @ Xc / (len(X) - 1)
vals, vecs = np.linalg.eigh(C)
order = np.argsort(vals)[::-1]
vals, vecs = vals[order], vecs[:, order]
print("covariance:\n", np.round(C, 3))
print("eigenvalues:", np.round(vals, 3), " explained ratio:", np.round(vals / vals.sum(), 3))
print("PC1 direction (sign may differ):", np.round(vecs[:, 0], 3))
print("scores on PC1:", np.round(Xc @ vecs[:, 0], 3))
````

### 2. Same result with scikit-learn

````python
# 2. Same result with scikit-learn
p = PCA(n_components=2).fit(X)
print("sklearn explained variance:", np.round(p.explained_variance_, 3))
````

### 3. Iris: how many components are enough?

````python
# 3. Iris: how many components are enough?
Z = StandardScaler().fit_transform(load_iris().data)
pca = PCA().fit(Z)
print("explained ratio:", np.round(pca.explained_variance_ratio_, 3))
print("cumulative:     ", np.round(np.cumsum(pca.explained_variance_ratio_), 3))
````

### 4. Reconstruction error with 2 components

````python
# 4. Reconstruction error with 2 components
p2 = PCA(n_components=2).fit(Z)
back = p2.inverse_transform(p2.transform(Z))
print("reconstruction MSE (2 comps):", round(np.mean((Z - back) ** 2) * Z.shape[1], 3),
      " = sum of discarded eigenvalues * (n-1)/n:", round(pca.explained_variance_[2:].sum() * (len(Z) - 1) / len(Z), 3))
````
