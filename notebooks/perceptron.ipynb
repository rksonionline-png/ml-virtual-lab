````python
import numpy as np
from sklearn.linear_model import Perceptron
````

### 1. The perceptron rule from scratch

````python
# 1. The perceptron rule from scratch
def train(X, y, eta=1.0, epochs=50):
    w, b = np.zeros(X.shape[1]), 0.0
    history = []
    for epoch in range(epochs):
        mistakes = 0
        for xi, yi in zip(X, y):
            if yi * (w @ xi + b) <= 0:                 # mistake: update
                w += eta * yi * xi
                b += eta * yi
                mistakes += 1
        history.append(mistakes)
        if mistakes == 0:
            break
    return w, b, history
````

### 2. A linearly separable data set

````python
# 2. A linearly separable data set
rng = np.random.default_rng(1)
X = np.vstack([rng.normal([2, 2], 0.6, (20, 2)), rng.normal([-2, -2], 0.6, (20, 2))])
y = np.array([1] * 20 + [-1] * 20)
w, b, hist = train(X, y)
print("separable data: mistakes per epoch", hist, " w =", np.round(w, 2), " b =", b)
print("training accuracy:", (np.sign(X @ w + b) == y).mean())
````

### 3. XOR is not linearly separable

````python
# 3. XOR is not linearly separable
Xx = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
yx = np.array([-1, 1, 1, -1])
_, _, hist = train(Xx, yx, epochs=20)
print("XOR: mistakes per epoch", hist[:10], "... never reaches 0")
````

### 4. Hand check of the first update (w = 0, b = 0, eta = 1, x = (2, 1), y = +1)

````python
# 4. Hand check of the first update (w = 0, b = 0, eta = 1, x = (2, 1), y = +1)
w0, b0 = np.zeros(2), 0.0
x, yv = np.array([2.0, 1.0]), 1
print("z =", w0 @ x + b0, " mistake:", yv * (w0 @ x + b0) <= 0, " new w =", w0 + yv * x, " new b =", b0 + yv)
````

### 5. scikit-learn's version

````python
# 5. scikit-learn's version
clf = Perceptron(eta0=1.0, max_iter=100, tol=None, random_state=0).fit(X, y)
print("sklearn perceptron accuracy:", clf.score(X, y))
````
