````python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
````

### Probability a sample appears in a bootstrap sample of size n

````python
# Probability a sample appears in a bootstrap sample of size n
for n in (10, 100, 10000):
    print(f"n = {n:5d}  P(in bootstrap) = {1 - (1 - 1/n)**n:.3f}")
print("Limit 1 - 1/e =", round(1 - np.exp(-1), 3), "(so ~36.8% are out-of-bag)")

X, y = load_breast_cancer(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)

tree = DecisionTreeClassifier(random_state=0).fit(Xtr, ytr)
print("Single tree test accuracy :", round(tree.score(Xte, yte), 3))

for n_trees in (1, 10, 50, 200):
    rf = RandomForestClassifier(n_estimators=n_trees, oob_score=n_trees >= 50,
                                random_state=0).fit(Xtr, ytr)
    oob = f"  OOB = {rf.oob_score_:.3f}" if n_trees >= 50 else ""
    print(f"{n_trees:3d} trees  test accuracy = {rf.score(Xte, yte):.3f}{oob}")

rf = RandomForestClassifier(n_estimators=200, random_state=0).fit(Xtr, ytr)
top = np.argsort(rf.feature_importances_)[::-1][:3]
print("Top-3 feature indices:", [int(i) for i in top],
      "importances:", np.round(rf.feature_importances_[top], 3))
````
