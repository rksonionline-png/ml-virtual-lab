````python
import numpy as np

rng = np.random.default_rng(0)
f = lambda x: 2.2 * np.sin(0.65 * x) + 0.25 * x - 1.2      # the unknown truth
xt = np.linspace(0.5, 9.5, 200)                               # grid for measuring bias and variance
sigma, n, runs = 0.8, 30, 300
````

### 1. Fit many training samples with polynomials of different degree

````python
# 1. Fit many training samples with polynomials of different degree
def fit_predict(deg):
    preds, train_err = [], []
    for _ in range(runs):
        x = rng.uniform(0, 10, n)
        y = f(x) + sigma * rng.normal(size=n)
        c = np.polyfit((x - 5) / 5, y, deg)
        preds.append(np.polyval(c, (xt - 5) / 5))
        train_err.append(np.mean((np.polyval(c, (x - 5) / 5) - y) ** 2))
    return np.array(preds), np.mean(train_err)
````

### 2. Bias-variance decomposition

````python
# 2. Bias-variance decomposition
print("deg  train_MSE  bias^2  variance  noise  bias^2+var+noise")
for deg in (1, 3, 5, 8):
    P, tr = fit_predict(deg)
    bias2 = np.mean((P.mean(axis=0) - f(xt)) ** 2)
    var = np.mean(P.var(axis=0))
    print(f"{deg:3d}  {tr:9.3f}  {bias2:6.3f}  {var:8.3f}  {sigma**2:5.2f}  {bias2 + var + sigma**2:8.3f}")
````

### 3. Finite-class generalisation bound

````python
# 3. Finite-class generalisation bound
H, eps, delta = 10**6, 0.05, 0.01
need = (np.log(H) + np.log(1 / delta)) / (2 * eps ** 2)
print("samples needed for |H| = 1e6, eps = 0.05, delta = 0.01:", int(np.ceil(need)))
for m in (500, 2000, 10000):
    print(f"n = {m:5d}  gap bound = {np.sqrt((np.log(H) + np.log(1 / delta)) / (2 * m)):.3f}")
````
