### 1. SVD of a small matrix

````python
# 1. SVD of a small matrix
A = np.array([[3.0, 1.0], [1.0, 2.0]])
U, s, Vt = np.linalg.svd(A)
print("singular values:", np.round(s, 4))
print("sqrt of eig(A^T A):", np.round(np.sqrt(np.sort(np.linalg.eigvalsh(A.T @ A))[::-1]), 4))
print("U S V^T equals A:", np.allclose(U @ np.diag(s) @ Vt, A))
````

### 2. Low-rank approximation of a structured 24 x 24 'image'

````python
# 2. Low-rank approximation of a structured 24 x 24 'image'
n = 24
x = (np.arange(n) + 0.5) / n
img = 0.5 + 0.35 * np.sin(14 * x)[None, :] + 0.15 * np.cos(9 * x)[:, None] * np.sin(4 * x)[None, :]
img = img + 0.12 * np.exp(-20 * (x[:, None] - x[None, :]) ** 2)    # a little extra structure
U, s, Vt = np.linalg.svd(img)
print("first 5 singular values:", np.round(s[:5], 3))
for k in (1, 2, 3, 5, 10):
    Ak = (U[:, :k] * s[:k]) @ Vt[:k]
    rel = np.linalg.norm(img - Ak) / np.linalg.norm(img)
    print(f"rank {k}: relative error = {rel:.4f}, numbers stored = {k * (2 * n + 1)} instead of {n * n}")
````

### 3. Eckart-Young: error of the best rank-k approximation equals sqrt(sum of dropped sigma^2)

````python
# 3. Eckart-Young: error of the best rank-k approximation equals sqrt(sum of dropped sigma^2)
k = 2
Ak = (U[:, :k] * s[:k]) @ Vt[:k]
print("||A - A_k||_F =", round(np.linalg.norm(img - Ak), 6), " sqrt(sum dropped sigma^2) =", round(np.sqrt((s[k:] ** 2).sum()), 6))
````

### 4. A random matrix has no low-rank structure

````python
# 4. A random matrix has no low-rank structure
R = np.random.default_rng(0).random((n, n))
sr = np.linalg.svd(R, compute_uv=False)
print("random matrix: energy in top 3 =", round((sr[:3] ** 2).sum() / (sr ** 2).sum(), 3),
      " structured image: ", round((s[:3] ** 2).sum() / (s ** 2).sum(), 3))
````
