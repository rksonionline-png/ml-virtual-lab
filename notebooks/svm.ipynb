````python
import numpy as np
from sklearn.svm import SVC
from sklearn.datasets import make_blobs, make_circles
````

### Linearly separable toy data

````python
# Linearly separable toy data
X = np.array([[1, 1], [2, 1], [1, 2], [4, 4], [5, 4], [4, 5]], dtype=float)
y = np.array([-1, -1, -1, 1, 1, 1])

clf = SVC(kernel="linear", C=1e6).fit(X, y)
w, b = clf.coef_[0], clf.intercept_[0]
print("w =", np.round(w, 3), " b =", round(b, 3))
print("Margin width 2/||w|| =", round(2 / np.linalg.norm(w), 3))
print("Support vectors:\n", clf.support_vectors_)
print("Functional margins y(w.x+b):", np.round(y * (X @ w + b), 3))
````

### Effect of C on noisy data

````python
# Effect of C on noisy data
Xn, yn = make_blobs(n_samples=100, centers=2, cluster_std=2.2, random_state=3)
for C in (0.01, 1, 100):
    m = SVC(kernel="linear", C=C).fit(Xn, yn)
    print(f"C = {C:6}: support vectors = {len(m.support_):3d}, train acc = {m.score(Xn, yn):.2f}")
````

### Kernel trick: concentric circles

````python
# Kernel trick: concentric circles
Xc, yc = make_circles(n_samples=200, factor=0.4, noise=0.08, random_state=0)
for kern in ("linear", "rbf"):
    print(f"{kern:6s} kernel accuracy on circles = {SVC(kernel=kern).fit(Xc, yc).score(Xc, yc):.2f}")
````
